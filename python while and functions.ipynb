    # While Loops & Functions

## PART 1: While Loops

### 1. Print "Hello" 3 times

```python
i = 1

while i <= 3:
    print("Hello")
    i += 1
```

### 2. Countdown from 10 to 1

```python
i = 10

while i >= 1:
    print(i)
    i -= 1
```

### 3. Odd numbers from 1 to 10

```python
i = 1

while i <= 10:
    if i % 2 != 0:
        print(i)
    i += 1
```

### 4. Table of 5

```python
i = 1

while i <= 5:
    print(f"5 x {i} = {5 * i}")
    i += 1
```

### 5. `break`: stop a loop early

```python
i = 1

while True:
    if i > 5:
        break
    print(i)
    i += 1
```

### 6. Countdown from 5 to 0

```python
x = 5

while x >= 0:
    print(x)
    x -= 1
```

### 7. Password check (`while True` + `break`)

```python
while True:
    password = input("Enter password: ")
    if password == "1234":
        break
    print("Wrong password, try again.")

print("Correct password!")
```

### 8. Even numbers from 0 to 10 (step of 2)

```python
count = 0

while count <= 10:
    print(count)
    count += 2
```

### 9. Loop through a string

```python
text = "Ali"
i = 0

while i < len(text):
    print(text[i])
    i += 1
```

### 10. `continue`: skip a value

```python
i = 0

while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

## PART 2: Functions

### 1. Add two numbers

```python
def add_numbers(a, b):
    return a + b

print(add_numbers(5, 3))
```

### 2. Even or odd

```python
def check_even_odd(n):
    if n % 2 == 0:
        return "Even"
    return "Odd"

print(check_even_odd(7))
```

### 3. Maximum of three numbers

```python
def find_maximum(a, b, c):
    return max(a, b, c)

print(find_maximum(10, 25, 15))
```

### 4. Area of a circle

```python
import math

def circle_area(radius):
    return math.pi * radius ** 2

print(round(circle_area(5), 2))
```

### 5. Celsius to Fahrenheit

```python
def celsius_to_fahrenheit(c):
    return (c * 9 / 5) + 32

print(celsius_to_fahrenheit(25))
```

### 6. Palindrome check (case-insensitive)

```python
def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]

print(is_palindrome("Madam"))
```

### 7. Square of a number

```python
def get_square(n):
    return n * n

print(get_square(6))
```

### 8. Factorial (with negative-number guard)

```python
def calculate_factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(calculate_factorial(5))
```

### 9. Count vowels

```python
def count_vowels(text):
    count = 0
    for char in text.lower():
        if char in "aeiou":
            count += 1
    return count

print(count_vowels("Hello World"))
```
